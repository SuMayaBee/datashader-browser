# NYC Taxi: 11.8 million pickups, instant zoom

Run the cell below, then **scroll to zoom** and **drag to pan**. Each view is re-aggregated from a precomputed [tile pyramid](tile-pyramid.ipynb) hosted next to this site, so a zoom fetches only the few tiles it needs instead of rescanning the 367 MiB source file.

In this browser, a full rescan of the source takes about 29 s per view. Reading tiles takes 0.03–0.33 s. Every level is identical to a native Datashader aggregate.

*Data: January 2015 NYC yellow taxi pickups, from the [HoloViz NYC Taxi example](https://examples.holoviz.org/gallery/nyc_taxi/nyc_taxi.html).*

````python
import datashader as ds
from datashader_browser import TilePyramid, site_url
from datashader_browser.holoviews import datashade

pyramid = TilePyramid(site_url("tiles/nyc-taxi"))
print(f"{pyramid.manifest['binned_rows']:,} pickups, zoom levels 0-{pyramid.max_level}")

datashade(pyramid, width=800, height=550, cmap="fire").opts(bgcolor="black")
````

## Use your own data

Build a pyramid once on your machine, then host the folder on any static server (GitHub Pages, S3, a plain web server):

```bash
pip install "datashader-browser[data]"
datashader-browser build-tiles points.parquet --x x --y y --max-level 7 --out tiles/
```

pyramid = TilePyramid("https://your-site.example/tiles")
