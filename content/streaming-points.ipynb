# Bounded-memory Datashader in the browser

This notebook generates chunks lazily and retains only one chunk plus the fixed-size aggregate.

```python
import numpy as np
import pandas as pd
import datashader as ds
import datashader.transfer_functions as tf
from datashader_browser import StreamingCanvas
```

```python
def point_chunks(total=2_000_000, chunk_size=100_000, seed=42):
    rng = np.random.default_rng(seed)
    for start in range(0, total, chunk_size):
        size = min(chunk_size, total - start)
        yield pd.DataFrame({
            'x': rng.normal(size=size),
            'y': rng.normal(size=size),
        })

canvas = StreamingCanvas(
    plot_width=800, plot_height=500,
    x_range=(-4, 4), y_range=(-4, 4),
)
aggregate = canvas.points(point_chunks(), 'x', 'y')
aggregate.attrs
```

```python
tf.shade(aggregate, cmap=['#101820', '#00a6a6', '#f2aa4c'], how='eq_hist').to_pil()
```
